# 03 — Evasion Strategy Classifier

Train a Random Forest to predict a pathogen's intracellular
survival strategy based on its effector features.

```python
from hostpathogen.ml.classifier import extract_features, train_classifier, plot_feature_importance
```

## Feature matrix

Each row is a pathogen. Features are derived from its effector repertoire.
The label (strategy) comes from the pathogens table.

```python
X, y = extract_features()
print("Features:")
display(X)
print("\nLabels:")
display(y)
```

## Train classifier

```python
model, report = train_classifier()
print("Training complete")
print()
import json
print(json.dumps(report, indent=2))
```

## Feature importance

Which features most influence the prediction?

```python
importances = plot_feature_importance(model)
for item in importances:
    print(f"{item['feature']:25s} {item['importance']:.4f}")
```
