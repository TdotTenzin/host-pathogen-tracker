# 02 — Interactome Analysis

Build and explore the host-pathogen interaction network.

```python
from hostpathogen.interactome import (
    build_network, hub_targets, network_stats,
    pathogen_subgraph, plot_network
)
```

## Build the network

```python
G = build_network()
stats = network_stats(G)
for k, v in stats.items():
    print(f"{k}: {v}")
```

## Top hub host proteins

These are the most-targeted host proteins across all pathogens.

```python
hubs = hub_targets(G, top_n=10)
for h in hubs:
    print(f"{h['host']:20s} degree={h['degree']:2d}  centrality={h['centrality']}")
```

## Subgraph: what does Salmonella target?

```python
sal_sub = pathogen_subgraph(G, "Salmonella enterica")
print(f"Salmonella subgraph: {sal_sub.number_of_nodes()} nodes, {sal_sub.number_of_edges()} edges")
for node in sal_sub.nodes():
    print(f"  {node}")
```

## Subgraph: what does Legionella target?

```python
leg_sub = pathogen_subgraph(G, "Legionella pneumophila")
print(f"Legionella subgraph: {leg_sub.number_of_nodes()} nodes, {leg_sub.number_of_edges()} edges")
for node in leg_sub.nodes():
    print(f"  {node}")
```

## Visualise the full network

```python
fig = plot_network(G, highlight_pathogen="Salmonella enterica")
fig.show()
```
