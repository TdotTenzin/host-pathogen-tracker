## PCA on host expression data

```python
from hostpathogen.ml.dimred import pca_analysis

pca = pca_analysis()
print("Explained variance ratio:", pca["explained_variance_ratio"])
print("Cumulative variance:", pca["cumulative_variance"])
print("Top genes PC1:", pca["top_genes_pc1"])
print("\nSample coordinates:")
for s in pca["samples"]:
    print(f"  {s['condition']:10s} PC1={s['PC1']:.2f} PC2={s['PC2']:.2f}")
```

## UMAP embedding

```python
from hostpathogen.ml.dimred import umap_analysis

umap = umap_analysis()
print("UMAP embedding:")
for s in umap["samples"]:
    print(f"  {s['condition']:10s} UMAP1={s['UMAP1']:.2f} UMAP2={s['UMAP2']:.2f}")
```

## PCA on pathogen features

How do pathogens cluster by their effector repertoire?

```python
from hostpathogen.ml.dimred import pathogen_feature_pca

ppca = pathogen_feature_pca()
print("Explained variance:", ppca["explained_variance_ratio"])
print("\nPathogen clustering:")
for s in ppca["samples"]:
    print(f"  {s['pathogen']:30s} {s['strategy']:25s} PC1={s['PC1']:.2f} PC2={s['PC2']:.2f}")
```

## Effector phylogenetics

Build a phylogenetic tree of all 25 effector proteins.

```python
from hostpathogen.ml.phylogenetics import build_phylogenetic_tree

tree = build_phylogenetic_tree()
print(f"Sequences: {tree['n_sequences']}")
print(f"Pathogens: {tree['n_pathogens']}")
print(f"Newick: {tree['newick'][:100]}...")
```

## Compare classifiers (LOOCV)

```python
from hostpathogen.ml.classifier import compare_classifiers

comparison = compare_classifiers()
print("Classifier comparison (Leave-One-Out CV):")
for c in comparison:
    print(f"  {c['model']:25s} accuracy={c['mean_accuracy']:.2f} +/- {c['std_accuracy']:.2f}")
```

## Cross-validation details

```python
from hostpathogen.ml.classifier import cross_validate_rf, grid_search_rf

cv = cross_validate_rf()
print(f"LOOCV accuracy: {cv['n_correct']}/{cv['n_folds']} = {cv['accuracy']:.2f}")
print("\nPer-fold results:")
for f in cv['folds']:
    mark = '+' if f['correct'] else 'X'
    print(f"  [{mark}] Fold {f['fold']}: {f['pathogen']:30s} true={f['true_strategy']:25s} pred={f['predicted_strategy']}")
```

## Grid search results

```python
gs = grid_search_rf()
print("Best params:", gs['best_params'])
print("Best score:", gs['best_score'])
print("\nAll results:")
for r in gs['all_results']:
    print(f"  {r['params']}  score={r['mean_score']:.2f}")
```
